```python
marks = np.array([40, 55, 62, 88, 95, 33, 71, 58])

marks = np.round(marks * 1.15, 2)
print("marks after 15% increase:", marks)
print()
marks = np.minimum(marks, 100)
print("corrected marks:", marks)
print()
print("students gained less than 10 marks:", np.sum(marks - np.array([40, 55, 62, 88, 95, 33, 71, 58]) < 10))
print()
original = np.array([40, 55, 62, 88, 95, 33, 71, 58])
print("students failing before:", np.sum(original < 40))
print("students failing after:", np.sum(marks < 40))
print()
print("percentage of topper marks:", np.round(original / np.max(original) * 100, 1))
print()
print("total marks given in revaluation:", np.sum(marks - original))
```

```python
marks = np.array([72, 45, 88, 95, 30, 67, 91, 54, 78, 39])

print("class average:", np.mean(marks))
print("middle value:", np.median(marks))
print()
print("gap between highest and lowest:", np.max(marks) - np.min(marks))
print()
print("index of topper:", np.argmax(marks))
print("index of weakest student:", np.argmin(marks))
print()
sorted_marks = np.sort(marks)
print("second highest marks:", sorted_marks[-2])
print()
print("students above class average:", np.sum(marks > np.mean(marks)))
print()
average = np.mean(marks)
std = np.std(marks)
print("students within one standard deviation:", np.sum((marks >= average - std) & (marks <= average + std)))
print()
print("class average after removing topper:", np.mean(marks[marks != np.max(marks)]))
```

```python
marks = np.array([[72, 85, 60, 91],
                  [45, 30, 98, 55],
                  [88, 66, 74, 100],
                  [95, 92, 55, 40]])

grace = np.array([5, 0, 8, 3])
weightage = np.array([1.2, 1.0, 1.1, 1.0])

marks = np.minimum(marks + grace, 100)
print("marks after grace:", marks)
print()
weighted = np.round(marks * weightage, 1)
print("weighted marks:", weighted)
print()
average = np.mean(marks, axis=0)
print("difference from subject average:", marks - average)
print()
bonus = np.array([5, 0, 2, 0])
print("marks after bonus:", np.array([[72, 85, 60, 91],
                                      [45, 30, 98, 55],
                                      [88, 66, 74, 100],
                                      [95, 92, 55, 40]]) + bonus)
print()
print("entries below subject average:", np.sum((marks - average) < 0, axis=0))
print()
spread = np.std(marks, axis=0)
print("standardized marks:", np.round((marks - average) / spread, 2))
```

```python
sec_A = np.array([[72, 85, 60],
                  [45, 30, 98],
                  [88, 66, 74]])

sec_B = np.array([[95, 92, 55],
                  [30, 48, 62]])

roll_C = np.array([201, 202, 203, 204])
marks_C = np.array([67, 82, 45, 91])
combined = np.vstack((sec_A, sec_B))

print("combined mark sheet:")
print(combined)
print("shape:", combined.shape)
print()
totals = np.sum(combined, axis=1)
combined_totals = np.hstack((combined, totals[:, np.newaxis]))
print("mark sheet with totals:")
print(combined_totals)
print()
roll_marks = np.hstack((roll_C[:, np.newaxis], marks_C[:, np.newaxis]))
print("roll numbers and marks:")
print(roll_marks)
print()
new_student = np.array([[205, 73]])
roll_marks = np.vstack((roll_marks, new_student))
print("after adding new student:")
print(roll_marks)
print()
serial = np.arange(1, 6)
serial = serial[:, np.newaxis]
combined_serial = np.hstack((serial, combined))
print("combined mark sheet with serial numbers:")
print(combined_serial)
print()
totals = np.sum(combined, axis=1)
above_200 = combined[totals > 200]
print("students whose total is above 200:")
print(above_200)
print()
grades = np.where(totals >= 225, "A",
                 np.where(totals >= 150, "B", "C"))
result = np.hstack((combined.astype(str), grades[:, np.newaxis]))
print("mark sheet with grades:")
print(result)
```

```python
import pandas as pd
import numpy as np
```

```python
##pd.DataFrame() - from a dictionary
restraunt_data = {
    'table_no':[12,14,15],
    'bill_amount':[200,300,400],
    'waiter_name':['Amit','Priya','Rahul']

}
df = pd.DataFrame(restraunt_data)
print(df)
print(type(df))
print(df['bill_amount'])
print(type(df['bill_amount']))
```

```python
#pd.Series()- 1D data with an index
total_bill = [4700,8000,5000,1200]
total_bill = pd.Series(total_bill, index = ['mon','tue','wed','fri'])
print(total_bill)
print(type(total_bill))

```

```python
#.values/.to_numpy()
print(df['bill_amount'].values)
print(type(df['bill_amount'].values))
print(df['bill_amount'].to_numpy())
print(type(df['bill_amount'].to_numpy()))
```

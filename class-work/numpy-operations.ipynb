```python
#artihmetic  - array with scalar , array with array
arr1 = np.arange(10)
print(arr1)
arr1= arr1+5
print(arr1)
print("Multiply",arr1*5)
print("subtraction",arr1-5) ## note: this all operation create a different array and do not change the original array
print("division",arr1/5)
print("exponent",arr1**2)


# array with array
x= np.arange(10)
y= np.arange(11,21)
print("x:",x)
print("y",y)
          # note: for this operation the length of both
          #       arrays should be same
z= x+y
print("sum of array x and y ", z)
print("product:",x*y)
```

```python
#other cases of array with array operations
x = np.arange(6).reshape(2,3)
y = np.arange(3).reshape(1,3)
print("x:",x)
print()
print("y:",y) # since this has different shape but it does not give error this happens due to broadcasting
              # the boradcasting tries to make the arrays of the same shape but even after trying it doesnt
              # have same shape it gives error
print()
print("x+y",x+y)
```

```python
#comparison operators - boolean arrays
x = np.arange(5)
y= np.arange(1,6)
print(x)
print(y)
print(x==y)
print(x>y)
print(x<y)
print(x!=y)
```

```python
# min ,max ,sum , mean , std
arr = np.arange(6).reshape(2,3)
print(arr)
print("Min",arr.min())
print("row min",arr.min(axis=1))
print("col min",arr.min(axis=0))
print()
print("Max",arr.max())
print("row max",arr.max(axis=1))
print("col max",arr.max(axis=0))
print()
print("sum of array ", arr.sum())
print("row sum",arr.sum(axis=1))
print("col sum",arr.sum(axis=0))
print()
print("mean of array ", arr.mean())
print("row mean",arr.mean(axis=1))
print("col mean",arr.mean(axis=0))
print()
print("median of array",np.median(arr))

```

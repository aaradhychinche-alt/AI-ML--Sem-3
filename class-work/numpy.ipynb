NUMPY Array


```python
import numpy as np
```

Creating an Array

```python
x = np.array([2,3,4,5])
x2 = np.array([[2,3],[3,4],[5,6]])
print(x)
print(x2)
print(type(x))
```

```python
#np.zeros ,np.ones(), np.full()
print(np.zeros(20))
print(np.zeros(10,dtype='int32'))
#we can also create 2d array using dimensions
print(np.zeros((3,4),dtype='int32')) #this creates a matrix of dimension 3 by 4

print(np.full((4,3),15)) # it fill the value 15 in the area of 4 by 3

```

```python
#np.arrange(start,stop,step)

print(np.arange(10,5,-2))

print(np.arange(10))

#np.reshape(dimensions)
print(np.arange(1,11).reshape(5,2))
```

```python
#np.linespace(start,stop,num)
print(np.linspace(1,10,5)) # in this it divides 1 to 10 in 5 equal parts
```

```python
#np.eye(n)
print(np.eye(4))# it createss an identity matrix of dimensions 4 by 4
print()
print(np.eye(3,4,k=2))# this creates idetity matrix of dimensions 3 by 4 and when the 3rd parameter is given like here k=2 it will start from index 2
```

```python
#.shape , n.dim ,.size , .dtype
x= np.arange(20).reshape(4,5)
print(x.shape)
print(x.ndim)
print(x.size)
print(x.dtype)
```

```python
#slicing - +ive and -ive
arr  = np.arange(10)
print(arr)
print(arr[-3:-12:-2])
```

```python
#slicing - submatrix (block-2D)
arr = np.arange(20).reshape(5,4)
print(arr)
print()
print(arr[2:,1:])
print()
print(arr[::4,::3])
print()
print(arr[1:3,2::-1])
```

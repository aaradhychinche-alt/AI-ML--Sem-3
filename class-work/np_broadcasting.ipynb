# Reshape


```python
#reshape(row,col)
arr = np.arange(12).reshape(4,-1)#when we  write -1 in col the python will it self decide how many columns to take based on the elements
print(arr)
## arr2= arr.reshape(5,-1) # this gives error becuase here the dimensions are not multiples
print(arr.shape)

#remember (10,) and (10,1) are different dimensions first is 1d and the latter is 2d
```

```python
arr = np.arange(10)
print(arr.shape)
arr1 = arr.reshape(1,-1)# look here we used -1 to make a row vector
print("ROW VECTOR",arr1)
arr2 = arr.reshape(-1,1) # look here we used -1 to make a column vector
print("COL VECTOR",arr2)
```

```python
## np.newaxis - can increase dimension to higher dimensions like(1d-to-2d),(2d-to-3d),etc.
arr1 = np.arange(12)
arr1 = arr1[:,np.newaxis]
print(arr1)
```

```python
#.ravil and .flatten
arr = np.arange(12).reshape(4,3)
print(arr)
arr1 = arr.ravel() # it make the shallow copy
arr1[0]= 99
arr2 = arr.flatten() # it makes deep copy
arr2 [2]= 100
print(arr1)
print(arr2)
```

```python
arr1 = np.ones(12).reshape(3,4)
arr1[0]=arr1[0]+10
arr1[1]=arr1[1]+20
arr1[2]=arr1[2]+30
print(arr1)
```

```python
#np.vstack() and np.hstack()
arr1 = np.array([[1,3,4],[2,0,5]])
arr2 = np.array([[5,6,7],[8,9,10]])
arr3 = np.vstack((arr1,arr2)) # in vstack the number of col should be same and it stacks vertically
arr4 = np.hstack((arr1,arr2))# in hstack the number of rows should be same and it stacks vertically
print("vstack:",arr3)
print()
print("hstack:",arr4)
```

```python
#np.concatenate()with axis
# when axis -> 0  it works as vstack
# and when axis ->1 it works as hstack

np.concatenate((arr1,arr2),axis=0)

np.concatenate((arr1,arr2),axis=1)
```

```python
#np.where(condition,true_val,false_val)
arr = np.array([2,-10,20,-15,-18,10])
arr1 = np.where(arr>0,arr,0)
print(arr1)

#nested np.where()
arr2 = np.array([20, 70, 45, 24, 100, 59])
arr3 = np.where(arr2>=80,"A",np.where((60<=arr2) & (arr2<=79),"B","C"))
print(arr3)
```

```python
marks = np.array([35,68,91,27,55])
print(np.where(marks>50))
print(marks[np.where(marks>50)])
print(np.where(marks>50)[0])
```

```python
# dot product
a = np.array([1,2,3])
b = np.array([4,5,6])
print("a*b",a*b)
print("np.dot",np.dot(a,b))
print("a@b", a@b)

a=np.array(([1,2,3],[4,5,6]))
#transpose
print(a.T)
print()
# to convert any matrix to square matrix multiply it too its transpose
print(a@a.T)
```
